```python
import endplay
from learning_tools import perform_bids
from tools import one_hot_hands, ohe_to_bid, mean_scores
from torch import load, tensor
import numpy as np
```

```python
#Environment
def bid_bridge(models):
    for Q in models:
        Q.eval()
    # Generate a deal
    deal = endplay.generate_deal()
    # Agents play
    bid_history = perform_bids(tensor(one_hot_hands(deal.north)).detach(),
                               tensor(one_hot_hands(deal.south)).detach(),
                               np.zeros(36), 
                               models,
                               5)
    #Compute reward by dummy analysis
    reward = mean_scores(deal.north, deal.south)[0]\
        [np.max(np.where(bid_history==1)[0])]
    
    #Print deal
    deal.east.clear()
    deal.west.clear()
    deal.pprint()
    #Return bids and final reward
    return ohe_to_bid(bid_history), reward
```

```python
# Agents
Q_models = []
Q_models.append(load("models/five_layers/l0.pth"))
Q_models.append(load("models/five_layers/l1.pth"))
Q_models.append(load("models/five_layers/l2.pth"))
Q_models.append(load("models/five_layers/l3.pth"))
Q_models.append(load("models/five_layers/l4.pth"))
```

```python
bid_bridge(Q_models)
```
